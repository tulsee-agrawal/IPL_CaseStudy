```python
import numpy as np
# Load CSV with possible missing values 
data = np.genfromtxt("data/deliveries.csv", delimiter=",", names=True, dtype=None, encoding="utf-8", missing_values="", filling_values=np.nan)
print(data)

# numpy load gen txt
matches_np = np.genfromtxt("data/matches.csv",delimiter=",",dtype=str,usecols=range(0,20))
print(matches_np[:5])

match_id = data['match_id']
inning = data['inning']
batsman_runs = data['batsman_runs']
total_runs = data['total_runs']
is_wicket = data['is_wicket']
over = data['over'].astype(int) 
batters = data['batter']
bowlers = data['bowler']
batting_teams = data['batting_team']
bowling_teams = data['bowling_team']
extras_type = data['extras_type']

print(type(match_id))
print(match_id.shape)
print(match_id[:5])
```

Task 1: Total Runs per Match

```python
matches = np.unique(match_id)

total_runs = np.array([
    np.sum(batsman_runs[match_id == m])
    for m in matches
])

result = [(int(m), int(r)) for m, r in zip(matches, total_runs)]

print(result[:5])
```

Task 2: Top 5 Batters

```python
players = np.unique(batters)

total_runs = np.array([
    np.sum(batsman_runs[batters == p])
    for p in players
])

sort = np.argsort(total_runs) #runs in ascending order, gives indices

top5_arr = sort[-5:][::-1] #last 5 

top5 = [(str(players[i]), int(total_runs[i])) for i in top5_arr]

print(top5)
```

Task 3: Strike Rate Calculation

```python
players = np.unique(batters)

total_runs = np.array([
    np.sum(batsman_runs[batters == p])
    for p in players
])

balls_faced = np.array([
    np.sum(batters == p)
    for p in players
])

strike_rate = (total_runs / balls_faced) * 100

result = [(str(players[i]), float(strike_rate[i])) for i in range(len(players))]

print(result[:5])
```

Task 4: Economy Rate of Bowlers

```python
unique_bowlers = np.unique(bowlers)

runs_conceded = np.array([
    np.sum(batsman_runs[bowlers == b])
    for b in unique_bowlers
])

balls = np.array([
    np.sum(bowlers == b)
    for b in unique_bowlers
])

overs = balls / 6

economy = runs_conceded / overs

result = [(str(unique_bowlers[i]), float(economy[i])) for i in range(len(unique_bowlers))]

print(result[:5])
```

Task 5: Runs per Over (1 to 20)

```python
avg_runs = []

for i in range(0, 20):
    total_runs = np.sum(batsman_runs[over == i])
    balls = np.sum(over == i)
    
    if balls > 0:
        avg = total_runs / (balls / 6)   
    else:
        avg = 0
        
    avg_runs.append(avg)

avg_runs = [round(float(x), 2) for x in avg_runs]
print(avg_runs)
```

Task 6: Boundary Analysis

```python
fours = np.sum(batsman_runs == 4)
sixes = np.sum(batsman_runs == 6)

print("Fours:", fours)
print("Sixes:", sixes)
```

```python
teams = np.unique(batting_teams)

boundary_counts = np.array([
    np.sum((batting_teams == t) & ((batsman_runs == 4) | (batsman_runs == 6)))
    for t in teams
])

top_team = teams[np.argmax(boundary_counts)]

print("Team with most boundaries:", top_team)
```

Task 7: Death Overs Analysis (Overs 16–20)

```python
death_mask = (over >= 16) & (over <= 20)

total_death_runs = np.sum(batsman_runs[death_mask])

print("Total runs in death overs:", total_death_runs)
```

```python
teams = np.unique(batting_teams)

team_runs = np.array([
    np.sum(batsman_runs[(batting_teams == t) & death_mask])
    for t in teams
])

top_team = teams[np.argmax(team_runs)]

print("Top team in death overs:", top_team)
```

Task 8: Highest Scoring Match

```python
matches = np.unique(match_id)

total_runs = np.array([
    np.sum(batsman_runs[match_id == m])
    for m in matches
])

max_idx = np.argmax(total_runs)

result = (int(matches[max_idx]), int(total_runs[max_idx]))

print(result)
```

Task 9: Match Winner Approximation


```python
matches = np.unique(match_id)

results = []

for m in matches:
    teams = np.unique(batting_teams[match_id == m])
    
    if len(teams) != 2:
        continue  
    
    t1, t2 = teams

    r1 = np.sum(batsman_runs[(match_id == m) & (batting_teams == t1)])
    r2 = np.sum(batsman_runs[(match_id == m) & (batting_teams == t2)])
    
    if r1 > r2:
        winner = t1
    elif r2 > r1:
        winner = t2
    else:
        winner = "Tie"
    
    results.append((int(m), str(winner)))

print(results[:5])
```

Task 10: Toss Impact Analysis

```python
match_ids_m = matches_np[1:, 0].astype(int)  
toss_winners = matches_np[1:, 9]              

toss_map = dict(zip(match_ids_m, toss_winners))
```

```python
matches = np.unique(match_id)

results = []

for m in matches:
    
    if m not in toss_map:
        continue
    
    toss_team = toss_map[m]
    
    teams = np.unique(batting_teams[match_id == m])
    
    if len(teams) != 2:
        continue
    
    t1, t2 = teams
    
    r1 = np.sum(batsman_runs[(match_id == m) & (batting_teams == t1)])
    r2 = np.sum(batsman_runs[(match_id == m) & (batting_teams == t2)])
    
    
    if toss_team == t1:
        toss_runs = r1
        opp_runs = r2
    else:
        toss_runs = r2
        opp_runs = r1
    
    if toss_runs > opp_runs:
        result = "Toss Winner Scored More"
    elif opp_runs > toss_runs:
        result = "Opponent Scored More"
    else:
        result = "Tie"
    
    results.append(result)
```

```python
results = np.array(results)

toss_more = np.sum(results == "Toss Winner Scored More")
opp_more = np.sum(results == "Opponent Scored More")
ties = np.sum(results == "Tie")

print("Toss team scored more:", toss_more)
print("Opponent scored more:", opp_more)
print("Ties:", ties)
```

Task 11: Match Scorecard Generation


```python
matches = np.unique(match_id)

for m in matches[:5]: 
    
    teams = np.unique(batting_teams[match_id == m])
    
    if len(teams) != 2:
        continue
    
    print(f"\nMatch {int(m)}:")
    
    for t in teams:
        runs = np.sum(batsman_runs[(match_id == m) & (batting_teams == t)])
        print(f" {t}: {runs} runs")
```
