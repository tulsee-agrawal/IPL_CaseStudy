```python
import numpy as np
# Load CSV with possible missing values 
data = np.genfromtxt("data/deliveries.csv", delimiter=",", names=True, dtype=None, encoding="utf-8", missing_values="", filling_values=np.nan)
print(data)

# numpy load gen txt
matches_np = np.genfromtxt("data/matches.csv",delimiter=",",dtype=str,usecols=range(0,20))
print(matches_np[:5])

match_id = data['match_id']
inning = data['inning']
batsman_runs = data['batsman_runs']
total_runs = data['total_runs']
is_wicket = data['is_wicket']
over = data['over'].astype(int) 
batters = data['batter']
bowlers = data['bowler']
batting_teams = data['batting_team']
bowling_teams = data['bowling_team']
extras_type = data['extras_type']

print(type(match_id))
print(match_id.shape)
print(match_id[:5])
```

Task 1: Total Runs per Match

```python
matches = np.unique(match_id)

total_runs = np.array([
    np.sum(batsman_runs[match_id == m])
    for m in matches
])

result = [(int(m), int(r)) for m, r in zip(matches, total_runs)]

print(result[:5])
```

Task 2: Top 5 Batters

```python
players = np.unique(batters)

total_runs = np.array([
    np.sum(batsman_runs[batters == p])
    for p in players
])

sort = np.argsort(total_runs) #runs in ascending order, gives indices

top5_arr = sort[-5:][::-1] #last 5 

top5 = [(str(players[i]), int(total_runs[i])) for i in top5_arr]

print(top5)
```

Task 3: Strike Rate Calculation

```python
players = np.unique(batters)

total_runs = np.array([
    np.sum(batsman_runs[batters == p])
    for p in players
])

balls_faced = np.array([
    np.sum(batters == p)
    for p in players
])

strike_rate = (total_runs / balls_faced) * 100

result = [(str(players[i]), float(strike_rate[i])) for i in range(len(players))]

print(result[:5])
```

Task 4: Economy Rate of Bowlers

```python
unique_bowlers = np.unique(bowlers)

runs_conceded = np.array([
    np.sum(batsman_runs[bowlers == b])
    for b in unique_bowlers
])

balls = np.array([
    np.sum(bowlers == b)
    for b in unique_bowlers
])

overs = balls / 6

economy = runs_conceded / overs

result = [(str(unique_bowlers[i]), float(economy[i])) for i in range(len(unique_bowlers))]

print(result[:5])
```

Task 5: Runs per Over (1 to 20)

```python
avg_runs = np.array([
    np.mean(batsman_runs[over == i]) if np.sum(over == i) > 0 else 0
    for i in range(1, 21)
])

print(avg_runs)
```
